```python
## With BG
import pickle
from src.sawmil.bag import Bag, BagDataset
from src.sawmil import sMIL, Linear, sAwMIL
from sklearn.preprocessing import StandardScaler
import numpy as np


with open('bd.pkl', 'rb') as f:
    data = pickle.load(f)
bags = [Bag(X=X, y=y) for X, y in zip(data[0], data[1])]
scaler = StandardScaler()
X = np.vstack([b.X for b in bags])
scaler.fit(X)
bt = [Bag(X=scaler.transform(b.X), y=b.y) for b in bags]
```

```python
bags = [Bag(X=X, y=y) for X, y in zip(data[0], data[1])]
solver_params = {
    #'env': {'LogFile': 'gurobi.log'},
    # 'model': {'Method': 2, 'Crossover': 0},
    #'start': np.zeros(train_ds.n_instances) # Specify a warm start (weights for each instance)
}
model = sAwMIL(C=1, kernel=Linear(), solver='gurobi', 
               normalizer='average',
               eta=0.1, verbose=True, solver_params=solver_params)
model.fit(BagDataset(bt[:100]), None)
```

```python
# load test
with open('bd_test.pkl', 'rb') as f:
    data_test = pickle.load(f)
bags_test = [Bag(X=scaler.transform(X), y=y) for X, y in zip(data_test[0], data_test[1])]
```

```python
yhat = model.predict(bags=bags_test)
from sklearn.metrics import matthews_corrcoef as mcc
```

```python
mcc([b.y for b in bags_test], yhat)
```
